```python
import torch
import numpy as np
from torch.utils.data import Dataset, DataLoader

                                                          
data = torch.load('dataset.pt')

                                                          
                                                                
split = np.load('train_val_test_split.npz')
train_indices = split['train_indices']
val_indices = split['val_indices']
test_indices = split['test_indices']

print(f"Učitano - Train: {len(train_indices)}, Val: {len(val_indices)}, Test: {len(test_indices)}")
```

```python
class DroneRFDataset(Dataset):
    """
    Custom PyTorch Dataset za drone RF signal dataset.
    Koristi spektrograme (x_spec) kao ulaz, i klasu (y) kao labelu.
    """

    def __init__(self, data_dict, indices):                                           
                                                                 
                                                                               
                                                            
                                                        
        self.x_spec = data_dict['x_spec']
        self.y = data_dict['y']
        self.indices = indices

    def __len__(self):
                                                                          
        return len(self.indices)

    def __getitem__(self, idx):
                                                                              
                                                                           
        real_idx = self.indices[idx]

        spectrogram = self.x_spec[real_idx]                        
        label = self.y[real_idx]                               

        return spectrogram, label

                                                                                                            
```

```python
                                                                       
train_dataset = DroneRFDataset(data, train_indices)
val_dataset = DroneRFDataset(data, val_indices)
test_dataset = DroneRFDataset(data, test_indices)

                                         
print(f"Broj uzoraka u train_dataset: {len(train_dataset)}")

                                                          
sample_spec, sample_label = train_dataset[0]
print(f"Oblik spektrograma: {sample_spec.shape}")
print(f"Labela: {sample_label} (tip: {type(sample_label)})")
```

```python
                                                                      
                                                                                                                           
                                                                                                                                
                                                                                                                            
                                                                

BATCH_SIZE = 64                                                                           

                                                                                 
                                              
train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=0                           
)

                                                                     
                                                                       
val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False)
test_loader = DataLoader(test_dataset, batch_size=BATCH_SIZE, shuffle=False)

                                                        
sample_batch_specs, sample_batch_labels = next(iter(train_loader))

print(f"Oblik batch-a spektrograma: {sample_batch_specs.shape}")
print(f"Oblik batch-a labela: {sample_batch_labels.shape}")
print(f"Labele u ovom batch-u: {sample_batch_labels}")
```
